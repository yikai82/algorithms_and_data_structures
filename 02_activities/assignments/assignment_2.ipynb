  # Question Two: Path to Leaves

  Given the `root` of a binary tree, return all root to leaf paths in any order.

  ## Examples

  ### Example 1

  ![](images/q1_ex1.png)

  Input: `root = [1, 2, 2, 3, 5, 6, 7]` *What traversal method is this?*

  Output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]]

````python
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

````

````python
# Input_1 = Tree_1
Tree_1 = TreeNode(1)
Tree_1.left = TreeNode(2)
Tree_1.left.left = TreeNode(3)
Tree_1.left.right = TreeNode(5)
Tree_1.right = TreeNode(2)
Tree_1.right.left = TreeNode(6)
Tree_1.right.right = TreeNode(7)
````

````python
# the output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]] is a path in a Pre-Order Tranversal

queue =[] 

def bt_path(root: TreeNode):  # -> List[List[int] 
    if root is None: 
        return []
    
    # base case as only one node input
    elif root.left is None and root.right is None:
        return[[root.val]]
    

    else: 
        path = []
        left_val = root.left.val
        if  left_val > root.val:    
            for node in bt_path(root.left):
                path.append([root.val] + node)
            
            for node in bt_path(root.right):
                path.append([root.val] + node) 
        else:
            for node in bt_path(root.right):
                path.append([root.val] + node) 

            for node in bt_path(root.left):
                path.append([root.val] + node)
    return path
    
````
