````python
def is_valid_brackets(s: str) -> bool:  # return will be Ture or False
    bracket_dict={')' : '(', 
                  ']' : '[',
                  '}' : '{'}

    stack=[]
    # for i in range(3): 
    for ch in s:
        if ch in bracket_dict.values(): # open bracket
            print(f"Push: {ch}")
            stack.append(ch)
            stack

        elif ch in bracket_dict.keys(): # close bracket
            print(f"Check: {ch} with the stack")
            if not stack: # check if stack is empty
                print('empty stack')
                return False
            
            if bracket_dict[ch] == stack[-1]:
                print(f"close meets open")
                stack.pop()

            else: 
                print(f"no match {ch}") 
    
    print(f"Stack: {stack}")
    return len(stack) == 0

            # if stack[-1] in bracket_dict.keys(): # close meets open
            #     # print(f"at {i-1}, and {i}, close meets open")
            #     # stack.pop()
````

````python
import random
import timeit
import pandas as pd
````

````python
def random_str_with_bracket_list(length: int, min_bracket: int):
    str_bracket ='()[]{}'
    str_combo = 'abcdefghijklmnopqrstuvwxyz0123456789!@#$%^&*_+-=:"<>?/'

    # 1. Set the number of bracket we want
    brackets = [random.choice(str_bracket) for _ in range(min_bracket)]
    
    # 2. Set the remaining  
    remain = [random.choice(str_combo) for _ in range(length - min_bracket)]
        
    # 3. combine
    result = brackets + remain
    random.shuffle(result)

    return ''.join(result)
````

````python
random.seed(123)
results = []

tests = [(20, 2), (200, 2), (1000, 2),
         (20, 4), (200, 20), (1000, 200),
         (20, 10), (200, 100), (1000, 100)]


for i, (length, brackets) in enumerate(tests, start=1): 
    s = random_str_with_bracket_list(length, brackets)
    valid = is_valid_brackets(s)
    t = timeit.timeit(lambda: is_valid_brackets(s), number=1000)
    results.append({
        "Test No.": i,
        "String": s,  # store the string itself
        "List Length": length,
        "Bracket #": brackets,
        "% Bracket": (brackets / length) * 100,
        'Valid': valid,
        "Time (sec)": t
    })
        
results_df = pd.DataFrame(results)


results_df

````
